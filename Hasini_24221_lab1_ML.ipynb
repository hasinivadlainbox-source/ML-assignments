```python
#Consider the given list as [2, 7, 4, 1, 3, 6]. Write a program to count pairs of elements with sum equal to 10.

a=[2,7,4,1,3,6];
count=0;
for i in range(len(a)):
    for j in range(i,len(a)):
        if (a[i]+a[j]==10):
            count+=1
print(count)
```

```python
#Write a program that takes a list of real numbers as input and returns the range (difference between minimum and maximum) of the list. Check for list 
#being less than 3 elements in which case return an error message (Ex: “Range determination not possible”). Given a list [5,3,8,1,0,4],the range is 8(8-0).

arr=list(map(float, input("enter the numbers list: ").split()))
if (len(arr) < 3):
    print("error: range determination not possible")
else:
    range= max(arr)-min(arr)
    print(range)
    print(max(arr) , "-" , min(arr))
```

```python
#Write a program that accepts a square matrix A and a positive integer m as arguments and returns A^m.
n = int(input("enter the order of the matrix: "))

A = []
print("enter the matrix: ")
for i in range(n):
    row = list(map(int, input().split()))
    A.append(row)

m = int(input("enter the power of the matrix: "))
        
result=[]
for i in range(n):
    row = []
    for j in range(n):
        if i==j:
            row.append(1)
        else:
            row.append(0)
    result.append(row)

for x in range(m):
    temp = []
    for i in range(n):
        row = []
        for j in range(n):
            s=0
            for k in range(n):
                s+=result[i][k]*A[k][j]
            row.append(s)
        temp.append(row)
    result = temp

print("Matrix raised to the power ", m , "is" )
for row in result:
    print(row)
```

```python
#Write a program to count the highest occurring character & its occurrence count in an input string. Consider only alphabets. Ex: for “hippopotamus” 
#as input string, the maximally occurring character is ‘p’ & occurrence count is 3.

s =str(input())
max_count = 0
character = s[0]
for i in s:
    count = s.count(i)
    if count > max_count:
        max_count = count
        character = i
print("highest occurring character : ", character)
print("Occurrence count: ", max_count)
```

```python
#Generate a list of 25 random numbers between 1 and 10. Find the mean, median and mode for these numbers.
import random

nums = []
for i in range(25):
    nums.append(random.randint(1,10))

print("random list: " , nums)
mean = sum(nums) / len(nums)
print("mean: ", mean)

sorted_nums = sorted(nums)
n=len(sorted_nums)
if (n%2==0):
    median = (sorted_nums[n//2-1]+sorted_nums[n//2])/2
else:
    median = sorted_nums[n//2]
print("median : ", median)

max_count=0
i=0
mode=nums[0]
for i in nums:
    count = nums.count(i)
    if count > max_count:
        max_count = count
        mode = i
print("mode : " , mode)
```
