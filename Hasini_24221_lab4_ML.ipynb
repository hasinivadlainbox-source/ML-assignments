```python
# A1: Modularized Functions for Lab 03 Experiments
import pandas as pd
import numpy as np

# GenAI Tool used: Gemini
def load_and_clean_data(filepath, sheet_name="marketing_campaign"):
    """Loads the dataset and returns the dataframe."""
    return pd.read_excel(filepath, sheet_name=sheet_name)

# GenAI Tool used: Gemini
def apply_label_encoding(df, column):
    """Applies label encoding to a specific categorical column."""
    df_encoded = df.copy()
    df_encoded[column] = df_encoded[column].astype('category').cat.codes
    return df_encoded

# GenAI Tool used: Gemini
def apply_one_hot_encoding(df, column):
    """Applies one-hot encoding to a specific categorical column."""
    return pd.get_dummies(df, columns=[column])

# GenAI Tool used: Gemini
def calculate_minkowski_distance(A, B, p):
    """Calculates Minkowski distance between two arrays A and B."""
    return np.sum(np.abs(np.array(A) - np.array(B)) ** p) ** (1 / p)

# GenAI Tool used: Gemini
def calculate_dot_product(A, B):
    """Calculates the dot product of two arrays."""
    return np.sum(np.array(A) * np.array(B))

# GenAI Tool used: Gemini
def calculate_statistics(data_array):
    """Calculates mean, variance, and standard deviation of a numeric array."""
    clean_data = np.array(data_array)[~np.isnan(data_array)]
    mean_val = np.sum(clean_data) / len(clean_data)
    variance_val = np.sum((clean_data - mean_val) ** 2) / len(clean_data)
    std_val = variance_val ** 0.5
    return mean_val, variance_val, std_val

# --- Execution ---
# Provide your local file path here
file_path = r"C:\Users\pc\Desktop\COLLEGE\SEM 5\ML\Lab Session Data.xlsx"
df = load_and_clean_data(file_path)

# Example execution matching Lab 3 variables
A = df[["MntWines", "MntFruits", "MntMeatProducts"]].iloc[0].values
B = df[["MntWines", "MntFruits", "MntMeatProducts"]].iloc[1].values
print(f"Minkowski Distance (p=2): {calculate_minkowski_distance(A, B, 2)}")
print(f"Dot Product: {calculate_dot_product(A, B)}")
```

```python
# A3: Performance comparison of 2 versions of k-means algorithm
import time
import numpy as np
from sklearn.cluster import KMeans

# 1. Manual Implementation (Representative of "Written by You")
def manual_kmeans(X, k, max_iters=100):
    # Randomly initialize centroids
    centroids = X[np.random.choice(X.shape[0], k, replace=False)]
    
    for _ in range(max_iters):
        # Calculate distances and assign clusters
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        labels = np.argmin(distances, axis=1)
        
        # Update centroids
        new_centroids = np.array([X[labels == i].mean(axis=0) if np.sum(labels == i) > 0 else centroids[i] for i in range(k)])
        
        # Check for convergence
        if np.all(centroids == new_centroids):
            break
        centroids = new_centroids
        
    return labels

# Generate synthetic data for stress testing the algorithms
np.random.seed(42)
# Simulating 5000 rows and 5 columns (similar to numerical features in your dataset)
X_test_data = np.random.rand(5000, 5) 

# --- Benchmark: Manual K-Means ---
start_time_manual = time.time()
manual_labels = manual_kmeans(X_test_data, k=3)
manual_execution_time = time.time() - start_time_manual

# --- Benchmark: AI/Scikit-Learn K-Means ---
# GenAI Tool used: Gemini
start_time_ai = time.time()
ai_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
ai_kmeans.fit(X_test_data)
ai_execution_time = time.time() - start_time_ai

print(f"Manual K-Means Execution Time: {manual_execution_time:.6f} seconds")
print(f"AI/Scikit-Learn K-Means Execution Time: {ai_execution_time:.6f} seconds")
print(f"Performance Difference: Scikit-Learn is {manual_execution_time / ai_execution_time:.2f}x faster.")
```
