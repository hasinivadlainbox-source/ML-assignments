```python
#A8. variance
import pandas as pd
import numpy as np
data = pd.read_excel(r"C:\Users\pc\Desktop\COLLEGE\SEM 5\ML\Lab Session Data.xlsx", sheet_name="marketing_campaign")
def my_mean(data):
    total = 0
    for x in data:
        total = total + x
    return total / len(data)
def my_variance(data):
    mean = my_mean(data)
    total = 0    
    for x in data:
        total = total + (x - mean) ** 2
    return total / len(data)
numeric_data = data.select_dtypes(include=np.number)
for column in numeric_data.columns:
    values = numeric_data[column].dropna()
    print(column, ":", my_variance(values))
```

```python
#A8. standard deviation
import pandas as pd
import numpy as np
data = pd.read_excel(r"C:\Users\pc\Desktop\COLLEGE\SEM 5\ML\Lab Session Data.xlsx", sheet_name="marketing_campaign")
def my_mean(data):
    total = 0
    for x in data:
        total = total + x
    return total / len(data)
def my_variance(data):
    mean = my_mean(data)
    total = 0
    for x in data:
        total = total + (x - mean) ** 2
    return total / len(data)
def my_std(data):
    return my_variance(data) ** 0.5
numeric_data = data.select_dtypes(include=np.number)
for column in numeric_data.columns:
    values = numeric_data[column].dropna()
    print(column, ":", my_std(values))
```

```python
#A11.
import pandas as pd
import numpy as np
data = pd.read_excel(r"C:\Users\pc\Desktop\COLLEGE\SEM 5\ML\Lab Session Data.xlsx", sheet_name="marketing_campaign")
data = data[["MntWines", "MntFruits", "MntMeatProducts"]].dropna()
data = data.values
def kmeans(data, k):
    centroids = data[:k].copy()
    while True:
        clusters = []
        for i in range(k):
            clusters.append([])
        for point in data:
            distances = []
            for centroid in centroids:
                distance = np.linalg.norm(point - centroid)
                distances.append(distance)
            nearest = distances.index(min(distances))
            clusters[nearest].append(point)
        new_centroids = []
        for cluster in clusters:
            new_centroids.append(np.mean(cluster, axis=0))
        new_centroids = np.array(new_centroids)
        if np.array_equal(centroids, new_centroids):
            break
        centroids = new_centroids
    return clusters, centroids
k = 3
clusters, centroids = kmeans(data, k)
for i in range(k):
    print("Cluster", i + 1)
    print(clusters[i])
print("\nCentroids:")
print(centroids)
```
